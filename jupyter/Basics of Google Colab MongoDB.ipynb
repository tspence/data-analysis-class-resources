This example notebook contains tutorials on the basics of Jupyter used within Google Colab.

Because Google Colab runs in the cloud, your Google Colab notebook may use a different IP address every time you launch it.  When working with MongoDB, it may be tricky to set up network rules to grant access to this server in a safe manner.

This notebook shows how to detect your Google Colab server's IP address so that you can grant network access to that machine for a short time.

***
# Fetch connection string from Google Colab secrets
***

```python
from google.colab import userdata
secret_key = userdata.get('mongodb')

# Remember not to print the actual connection string! 
# You don't want to share it accidentally.
print(f"My MongoDB connection string is {len(secret_key)} characters.")
```

***
# Determine Google Colab IP address
***

```python
# Tutorial from https://stackoverflow.com/questions/2311510/getting-a-machines-external-ip-address-with-python

# This example requires the requests library be installed.  You can learn more
# about the Requests library here: http://docs.python-requests.org/en/latest/
from requests import get

ip = get('https://api.ipify.org').content.decode('utf8')
print(f"My Google Colab notebook's IP address is: {ip}")
print(f"If you have problems connecting to MongoDB in the next code block, check")
print(f"the Network Access tab in the MongoDB Atlas website and ensure this IP")
print(f"address has been granted access.")
```
